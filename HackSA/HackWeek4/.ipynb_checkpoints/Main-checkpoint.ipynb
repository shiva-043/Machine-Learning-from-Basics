```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_initial = pd.read_csv('Train_dataset.csv')
test_initial = pd.read_csv('Test_dataset.csv')
```

```python
train_initial.head(5)
```

```python
print(train_initial.isna().sum()/len(train_initial))
print(train_initial.isna().sum(),len(train_initial))
```

```python
#dropping nan target rows
train_initial.dropna(subset = ['age_group'],inplace = True)
```

```python
train_initial['age_group'] = train_initial['age_group'].map({
    'Adult':0,
    'Senior':1
})
```

```python
train_initial.sample(5)
```

```python
# Lets analyse the dataaa
```

```python
for col in train_initial:
    print(f"\n{col}:")
    print(train_initial[col].value_counts(dropna=False))
```

```python
# Gender is 1 and 2 - binary good if 0 and 1
# PAQ605 has value 7.0 differner - all are 1 and 2 - lets remocve it
# DIQ010 has value 3.0 differnet-- learnt it means borderline 0.5
```

```python
train_initial["PAQ605"] = train_initial["PAQ605"].replace({7.0: np.nan})
test_initial["PAQ605"] = test_initial["PAQ605"].replace({7.0: np.nan})

train_initial["PAQ605"] = train_initial["PAQ605"].map({1.0: 1, 2.0: 0})
test_initial["PAQ605"] = test_initial["PAQ605"].map({1.0: 1, 2.0: 0})

train_initial["DIQ010"] = train_initial["DIQ010"].replace({1.0: 1, 2.0: 0, 3.0: 0.5})
test_initial["DIQ010"] = test_initial["DIQ010"].replace({1.0: 1, 2.0: 0, 3.0: 0.5})

train_initial["RIAGENDR"] = train_initial["RIAGENDR"].map({1.0: 0, 2.0: 1})
test_initial["RIAGENDR"] = test_initial["RIAGENDR"].map({1.0: 0, 2.0: 1})
```

```python
#classs weight checking
display(train_initial['age_group'].value_counts()) #imbalanced , more 0 

plt.figure(figsize = (4,3))
sns.countplot(data = train_initial, x = 'age_group')
plt.show()
```

```python
plt.figure(figsize = (8,6))
correlation_matrix = train_initial.corr()
sns.heatmap(correlation_matrix,annot = True,cmap = 'coolwarm',fmt = '.2f')
plt.show()
```

```python
features = ["RIAGENDR",	"PAQ605" ,"BMXBMI",	"LBXGLU","DIQ010","LBXGLT","LBXIN"]
numerical_cols = ["BMXBMI", "LBXGLU", "LBXGLT", "LBXIN"]
binary_cols    = ["RIAGENDR", "PAQ605", "DIQ010"]
```

```python
X = train_initial.drop(columns = ["SEQN","age_group"])
test = test_initial.drop(columns = ["SEQN"])

y = train_initial['age_group']
```

```python
# Missing values handling
```

```python
from sklearn.impute import SimpleImputer
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("impute", SimpleImputer(strategy="median"))#,
        # ("scale", StandardScaler())
    ]), numerical_cols),
    
    ("bin", SimpleImputer(strategy="most_frequent"), binary_cols)
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
```

```python
#best score checking 
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
models = [
    LogisticRegression(class_weight="balanced", random_state=43, max_iter=1000),
    RandomForestClassifier(class_weight="balanced", random_state=43),
    LGBMClassifier(class_weight="balanced", random_state=43, verbose=-1),
    XGBClassifier(random_state=43,eval_metric="logloss",scale_pos_weight=1638/314),
    CatBoostClassifier(random_state=43,verbose=0,auto_class_weights="Balanced")
]
```

```python
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
pipelines = [
    Pipeline([
        ("pre", preprocessor), 
        # ('smote',SMOTE(random_state = 43)),
        ("clf", model)
    ])
    for model in models
]
scores = [cross_val_score(pipeline,X,y,cv=cv,scoring="f1_macro") for pipeline in pipelines ]
```

```python
for score in scores:
    print(score.mean())
```

```python
lgbm = Pipeline([
    ("preprocessor", preprocessor),
    ('smote',SMOTE(random_state = 43)),
    ("classifier", LGBMClassifier(
        random_state=43,
        verbosity=-1,
        scale_pos_weight=1638/314
    ))
])

param_dist = {
    "classifier__n_estimators": [250, 275, 300, 325, 350],
    "classifier__learning_rate": [0.04, 0.045, 0.05, 0.055, 0.06],

    "classifier__max_depth": [4, 5, 6],
    "classifier__num_leaves": [7, 8, 9, 10],

    "classifier__min_child_samples": [8, 9, 10, 11, 12],

    "classifier__subsample": [0.85, 0.9, 0.95],
    "classifier__colsample_bytree": [0.85, 0.9, 0.95],

    "classifier__reg_alpha": [0.0, 0.05, 0.1, 0.15],
    "classifier__reg_lambda": [2, 3, 4],

    "classifier__min_split_gain": [0.0, 0.01, 0.02]
}

random_search = RandomizedSearchCV(
    estimator=lgbm,
    param_distributions=param_dist,
    n_iter=50,              
    scoring="f1_macro",
    random_state=43,
    n_jobs=-1,
    verbose=0
)

random_search.fit(X, y)
```

```python
# print(random_search.best_score_)
# print(random_search.best_params_)
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import f1_score
import numpy as np

probs = cross_val_predict(
    random_search.best_estimator_,
    X,
    y,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

thresholds = np.arange(0.10, 0.91, 0.01)

best_threshold = 0.5
best_score = -1

for t in thresholds:
    preds = (probs >= t).astype(int)
    score = f1_score(y, preds,average = "macro")

    if score > best_score:
        best_score = score
        best_threshold = t

print("Best threshold:", best_threshold)
print("Best F1:", best_score)
```

```python
lgbm_best = random_search.best_estimator_

# Train on full training data
lgbm_best.fit(X, y)

# Get probabilities
probs = lgbm_best.predict_proba(test)[:, 1]

# Apply your optimized threshold
pred = (probs >= 0.56).astype(int)

submission = pd.DataFrame({
    "age_group": pred
})

submission.to_csv("submission.csv", index=False)
```
